```python
#### PCR



import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib as mpl


# Update rcParams for consistent formatting
plt.rcParams.update({
    'figure.dpi': 300,
    'savefig.dpi': 300,
    'font.size': 6,
    'axes.titlesize': 6,
    'axes.labelsize': 6,
    'xtick.labelsize': 6,
    'ytick.labelsize': 6,
    'legend.fontsize': 6
})

mpl.rcParams['font.family'] = 'DejaVu Sans'  # or 'Arial'
mpl.rcParams['svg.fonttype'] = 'none'


# Sample DataFrame
data = pd.DataFrame({
    'Group': ['U7ex51', 'U7ex51', 'U7ex51', 'ASO', 'ASO', 'ASO',
              'U7ex51', 'U7ex51', 'U7ex51', 'ASO', 'ASO', 'ASO'],
    'Value': [49.1, 58.6, 69.2,22, 20.4, 55.5, 85, 62.2, 67.5,62.5,8.6,29.3],
    'Region': ['Hippocampus', 'Hippocampus', 'Hippocampus', 'Hippocampus', 'Hippocampus', 'Hippocampus',
               'Cortex', 'Cortex', 'Cortex', 'Cortex', 'Cortex', 'Cortex']
})


# Set category order: ASO first
data['Group'] = pd.Categorical(data['Group'], categories=['ASO', 'U7ex51'], ordered=True)

# Define bar color palette
bar_palette = {
    "ASO":"#8DA0CB",
    "U7ex51":"#E78AC3"
}

# Plot
plt.figure(figsize=(4,3))

# Barplot: mean ± SD per group in each region
sns.barplot(
    data=data,
    x='Region',
    y='Value',
    hue='Group',
    palette=bar_palette,
    errorbar="se",
    edgecolor='black',
    dodge=True
)

# Stripplot: uses hue for correct dodge, but forces black dots
sns.stripplot(
    data=data,
    x='Region',
    y='Value',
    hue='Group',        # needed for dodge alignment
    dodge=True,
    palette=['black'],  # forces all dots to be black
    jitter=True,
    size=6,
    alpha=0.7,
    marker='o'
)

# Fix the legend to show only the bars (first 2 handles)
handles, labels = plt.gca().get_legend_handles_labels()
n_groups = data['Group'].nunique()
plt.legend(handles[:n_groups], labels[:n_groups], title='Group')

# Style
sns.despine()
plt.tight_layout()

plt.ylabel("Exon skipping rate (%)", fontsize=10)
plt.xlabel("Region", fontsize=10)

# Adjust tick label font size
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

# Optional: legend font size##plt.legend(title='Group', fontsize=10, title_fontsize=10)

# Save as SVG with 300 dpi
plt.savefig('/exports/humgen/qmao/grouped_bar_dot_plot.svg', dpi=300, format='svg')

# Show plot
plt.show()
```
