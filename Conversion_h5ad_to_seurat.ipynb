```python
import scanpy as sc 
from scipy import io
```

```python
!mkdir matrix_files3
```

```python
adata = sc.read_h5ad('/exports/archive/hg-exon-skip/alireza/expression_matrices/file_sampled.h5ad')
```

```python
with open('matrix_files3/barcodes.tsv' , 'w') as f:
    for item in adata.obs_names:
        f.write(item +'\n')
```

```python
with open('matrix_files3/features.tsv', 'w') as f:
    for item in ['\t'.join([x,x,'Gene Expression']) for x in adata.var_names]:
        f.write(item + '\n')
```

```python
io.mmwrite('matrix_files3/matrix', adata.X.T)
```
